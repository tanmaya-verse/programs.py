```python
nums = [3, 4, 7, 8, 10, 12, 15]
evens = []
squares = []
big = []
for n in nums:
    if n % 2 == 0:
        evens.append(n)
for e in evens:
    squares.append(e * e)
for s in squares:
    if s > 100:
        big.append(s)
print(evens, squares, big)
```

```python
words = ["cat", "tree", "apple", "sun", "book"]
result = []
for w in words:
    if len(w) > 3:
        result.append(w.upper())
print(result)
```

```python
nums = [1, 2, 3, 4, 5, 6, 7, 8, 9]
result = []
for n in nums:
    if n % 2 == 0 or n % 3 == 0:
        result.append(n * n)
print(result)
```

```python
nums = [5, 15, 20, 25, 30]
added = []
result = []
for n in nums:
    added.append(n + 10)
for a in added:
    if a > 30:
        result.append(a)
print(added, result)
```

```python
nums = [-5, 3, -8, 10, -1]
result = []
for n in nums:
    if n < 0:
        result.append(abs(n))
print(result)
```

```python
words = ["train", "apple", "moon", "pen", "sky"]
result = []
for w in words:
    if w.endswith("n"):
        result.append(len(w))
print(result)
```
