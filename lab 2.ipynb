```python
#6
text = input("Enter a string: ")

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)
```

```python
#5
students = {}

data = input("Enter 5 student names and marks: ").split()

for i in range(0, 10, 2):
    name = data[i]
    marks = int(data[i + 1])
    students[name] = marks

print(students)
```
