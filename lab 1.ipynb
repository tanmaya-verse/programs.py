```python
# 2 star pattern right 
row = int(input("Enter a number:"))
for i in range(1, row+1):
    print("*"*i)
```

```python
# 3 star pattern inverted 
row = int(input("Enter a number:"))
for i in range(row, 0, -1):
    print("*"*i)
```

```python
# 4 star pattern pyramid
row = int(input("Enter a number:"))
for i in range(1, row+1):
    space = " " * (row-i)
    star = "*" * (2*i-1)
    print(space + star)
```

```python
# fibonacci series
n = int(input("Enter number of terms: "))
a = 0
b = 1
print("Fibonacci Series:")
for i in range(n):
  print(a, end=" ")
  c = a + b
  a = b
  b = c
```

```python
# 6 
text = "Dr. BabaSaheb Satpute"

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print("Character frequencies:", frequency)
```

```python
balance = 0
history = []

while True:
    print("\n1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Transaction History")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == '1':
        amount = float(input("Enter amount: "))
        if amount > 0:
            balance = balance + amount
            history.append(("Deposit", amount, balance))
            print("Deposit successful")
        else:
            print("Invalid amount")

    elif choice == '2':
        amount = float(input("Enter amount: "))
        if amount <= 0:
            print("Invalid amount")
        elif amount > balance:
            print("Insufficient balance")
        else:
            balance = balance - amount
            history.append(("Withdraw", amount, balance))
            print("Withdrawal successful")

    elif choice == '3':
        print("Balance:", balance)

    elif choice == '4':
        for t in history:
            print(t)
        if len(history) == 0:
            print("No transactions")

    elif choice == '5':
        break

    else:
        print("Invalid choice")
```
