```python
# 1. 

nums = list(map(int, input("Enter integers: ").split()))
squares = [x * x for x in nums if x % 2 == 0]
result = [s for s in squares if s > 100]
print(result)
```

```python
2.

words = input("Enter strings: ").split()
result = [w.upper() for w in words if len(w) > 3]
print(result)
```

```python
3.

nums = list(map(int, input("Enter integers: ").split()))
result = [x * x for x in nums if x % 2 == 0 or x % 3 == 0]
print(result)
```

```python
4.

nums = list(map(int, input("Enter integers: ").split()))
added = [x + 10 for x in nums]
result = [x for x in added if x > 30]
print(result)
```

```python
6.

words = input("Enter words: ").split()
result = [len(w) for w in words if w.endswith('n')]
print(result)
```
